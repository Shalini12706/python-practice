```python
#Converting obj-Numerical
cols = ['fuel', 'seller_type', 'transmission', 'owner']
```

```python
df = pd.get_dummies(df, columns=cols, dtype=int)
```

```python
df.info()
```

```python
df['seats'] = df['seats'].fillna(df['seats'].mean())
```

```python
df.isnull().sum()
```

```python
#outlier treatment
sns.boxplot(df[['selling_price', 'km_driven']])
plt.title("Outlier Detection (Before Treatment)")
plt.show()
```

```python
for col in ['selling_price', 'km_driven']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[col] = np.where(df[col] > upper, upper, np.where(df[col] < lower, lower, df[col]))
```

```python
#Outlier treatment
sns.boxplot(df[['selling_price', 'km_driven']])
plt.title("Outliers Handled (After Treatment)")
plt.show()
```
